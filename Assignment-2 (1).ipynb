```python
#Answer of question 03

missing_values = wdi[columns].isnull().sum()
max_missing_var = missing_values.idxmax()
max_missing_count = missing_values.max()
print(f"3. Variable with highest missing values: '{max_missing_var}' with {max_missing_count} missing values")
```

```python
#Answer of question 05

gdp_growth_data = wdi[['Country','Year','GDP per capita growth (annual %)']].copy()
first_year = wdi['Year'].min()
last_year = wdi['Year'].max()


first_gdp = gdp_growth_data[gdp_growth_data['Year'] == first_year].set_index('Country')['GDP per capita growth (annual %)']
last_gdp = gdp_growth_data[gdp_growth_data['Year'] == last_year].set_index('Country')['GDP per capita growth (annual %)']

avg_growth = gdp_growth_data.groupby('Country')['GDP per capita growth (annual %)'].mean()
overall_avg_growth = avg_growth.mean()

print(f"5. Average GDP per capita growth rate: {overall_avg_growth:.2f}%")
```
